```python
import json, pandas as pd
h = json.load(open('site/hist.json'))
smc = pd.DataFrame(h['SMC']).T.astype(float)
smc.columns = ['orcado','atualizado','empenhado','liquidado','pago']
smc.index = smc.index.astype(int)
smc['atualizado/orcado'] = smc.atualizado/smc.orcado
smc['pago/atualizado'] = smc.pago/smc.atualizado
smc.round(2).tail(10)
```

```python
ax = (smc[['orcado','atualizado','empenhado','pago']]/1e6).plot(figsize=(9,4), title='SMC (órgão 25) — R$ milhões nominais')
ax.set_xlabel('');
```
